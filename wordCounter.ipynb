```python
import gradio as gr
```

```python
def count_it(text):

    words = len(text.split())
    characters = len(text)

    if text.split():
        longest_word = max(text.split(), key=len)
    else:
        longest_word = ""

    return (
        f"{words} words, {characters} characters",
        longest_word
    )
```

```python
print(count_it("Hello everyone welcome"))
```

```python
demo = gr.Interface(
    fn=count_it,
    inputs=gr.Textbox(label="Enter Text"),
    outputs=[
        gr.Textbox(label="Word Count"),
        gr.Textbox(label="Longest Word")
    ],
    title="Word Counter",
    description="Count words, characters, and find the longest word."
)

demo.launch()
```
