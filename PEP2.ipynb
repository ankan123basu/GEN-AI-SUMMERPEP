```python
from pydantic import BaseModel, Field
from typing import TypedDict
```

```python
class Student(BaseModel):
    #  name: str
    #  age: int
    name : str = Field(max_length=50, description="The name of the student")
    age : int = Field(lt= 150, gt=0, description="The age of the student, must be less than or equal to 150")
# class Student(TypedDict):
#     name: str
#     age: int
```

```python
# def greet(name, age):
#     return "Hello " + name + "!" + " You are " + str(age) + " years old."
# greet("abcd", 25)

from litellm import completion


def greet(name, age):
    response = completion(
        model="openai/gpt-4o-mini",
        messages=[
            {"role": "user", "content": f"Hello, my name is {name} and I am {age} years old. Can you greet me?"}
        ]
    )
    return response.choices[0].message.content

greet("abcd", 25)
```

```python
gr.Interface(fn=greet, inputs=[gr.Textbox(label="Name"), gr.Number(label="Age")], outputs=gr.Textbox(label="Greeting")).launch()
```
