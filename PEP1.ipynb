```python
from getpass import getpass
open_api_key = getpass("Enter OpenAI api Key")
```

```python
gemini_api_key = getpass("Enter Gemini api Key")
```

```python
openai_client = OpenAI(api_key=open_api_key)
```

```python
gemini_client = OpenAI(api_key=gemini_api_key,base_url="https://generativelanguage.googleapis.com/v1beta/openai/")
```

```python
response = gemini_client.chat.completions.create(
    # model="gpt-4o-mini",
    model = "gemini-3.6-flash",
    messages=[
        {"role": "system", "content": "You are a helpful assistant, you generate 3 taglines in emojis for a given company."},
        {"role": "user", "content": "Google"},
    ]
)
print(response.choices[0].message.content)
# print(response.model_dump_json(indent=2))
```
